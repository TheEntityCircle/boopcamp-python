## 1. Импорт NumPy

Обычно библиотеку импортируют под коротким именем `np`.

```python
import numpy as np

print("numpy version:", np.__version__)
```

## 2. Создание массивов

Главный объект NumPy — `ndarray`.

Самый простой способ создать массив — передать список в `np.array()`.

```python
a = np.array([10, 20, 30, 40, 50])

print(a)
print(type(a))
```

У массива есть полезные свойства:

- `a.size` — количество элементов;
- `a.ndim` — количество измерений;
- `a.dtype` — тип элементов.

```python
print("size:", a.size)
print("ndim:", a.ndim)
print("dtype:", a.dtype)
```

### Несколько удобных способов создать массив

`np.arange()` похож на обычный `range()`.

`np.linspace()` создаёт заданное количество равномерно расположенных чисел.

```python
print("arange:", np.arange(0, 10, 2))
print("linspace:", np.linspace(0, 1, 5))
print("zeros:", np.zeros(5))
print("ones:", np.ones(5))
```

## 3. Типы данных

В отличие от обычного списка Python, NumPy-массив обычно хранит элементы **одного типа**.

Тип можно посмотреть через `dtype`.

```python
integers = np.array([1, 2, 3])
floats = np.array([1.0, 2.5, 3.7])

print(integers.dtype)
print(floats.dtype)
```

Тип можно указать явно.

```python
a = np.array([1, 2, 3], dtype=float)

print(a)
print(a.dtype)
```

## 4. Индексация и срезы

Индексация очень похожа на работу со списками Python.

```python
a = np.array([10, 20, 30, 40, 50])

print("Первый:", a[0])
print("Последний:", a[-1])
print("Срез:", a[1:4])
print("Каждый второй:", a[::2])
```

Элементы можно изменять.

```python
a = np.array([10, 20, 30, 40, 50])

a[0] = 100
a[2:4] = 0

print(a)
```

## 5. Операции над массивами

Одна из главных идей NumPy — выполнять операцию **сразу над всем массивом**.

В обычном Python для этого часто используют цикл или list comprehension.

```python
numbers = [1, 2, 3, 4, 5]

python_result = [x * 2 for x in numbers]

print(python_result)
```

В NumPy можно написать короче:

```python
numbers = np.array([1, 2, 3, 4, 5])

print(numbers * 2)
print(numbers + 10)
print(numbers - 1)
print(numbers ** 2)
```

Такие операции выполняются **поэлементно**.

Можно использовать и готовые математические функции:

```python
numbers = np.array([1, 4, 9, 16, 25])

print("sqrt:", np.sqrt(numbers))
```

## 6. Простые встроенные функции

NumPy содержит много готовых функций для обработки числовых данных.

```python
data = np.array([7, 2, 9, 4, 6, 1])

print("Сумма:", np.sum(data))
print("Среднее:", np.mean(data))
print("Минимум:", np.min(data))
print("Максимум:", np.max(data))
```

Иногда нужен не сам минимум или максимум, а **его индекс**.

```python
data = np.array([7, 2, 9, 4, 6, 1])

print("Индекс минимума:", np.argmin(data))
print("Индекс максимума:", np.argmax(data))
```

Ещё несколько несложных функций:

```python
data = np.array([4, 2, 7, 2, 9, 4, 2])

print("sort:", np.sort(data))
print("unique:", np.unique(data))
print("round:", np.round(np.array([1.234, 5.678, 9.876]), 1))
```

## 7. Сравнения

Если сравнить массив с числом, NumPy сравнит с ним **каждый элемент**.

```python
a = np.array([1, 5, 8, 2, 10, 3])

print(a > 4)
```

Получаем булевый массив из `True` и `False`.

Можно строить более сложные условия.

Для массивов используются:

- `&` — И
- `|` — ИЛИ
- `~` — НЕ

Каждое отдельное сравнение лучше брать в скобки.

```python
a = np.array([1, 5, 8, 2, 10, 3])

condition = (a >= 3) & (a <= 8)

print(condition)
```

## 8. Фильтрация

Булевый массив можно использовать как **маску**.

NumPy оставит только те элементы, для которых условие равно `True`.

```python
a = np.array([1, 5, 8, 2, 10, 3])

print("Больше 4:", a[a > 4])
print("Чётные:", a[a % 2 == 0])
print("От 3 до 8:", a[(a >= 3) & (a <= 8)])
```

По маске можно не только получать значения, но и изменять их.

```python
a = np.array([1, 5, 8, 2, 10, 3])

a[a > 5] = 0

print(a)
```

## 9. `np.any()` и `np.all()`

`np.any()` проверяет, есть ли **хотя бы один** `True`.

`np.all()` проверяет, являются ли **все** значения `True`.

```python
a = np.array([2, 4, 6, 8])

print("Есть число > 5:", np.any(a > 5))
print("Все положительные:", np.all(a > 0))
print("Все чётные:", np.all(a % 2 == 0))
```

## 10. Сравнение скорости с обычным Python

Теперь сравним обычный Python и NumPy на большой последовательности чисел.

Для измерения времени используем `time.perf_counter()`.

```python
from time import perf_counter

N = 10_000_000

python_data = list(range(N))
numpy_data = np.arange(N)
```

```python
start = perf_counter()

python_result = [x ** 0.5 for x in python_data]

python_time = perf_counter() - start

print(f"Обычный Python: {python_time:.6f} секунд")
```

```python
start = perf_counter()

numpy_result = np.sqrt(numpy_data)

numpy_time = perf_counter() - start

print(f"NumPy: {numpy_time:.6f} секунд")
```

```python
print(f"NumPy быстрее примерно в {python_time / numpy_time:.1f} раз")
```
