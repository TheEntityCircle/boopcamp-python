## 09_loops.py

````python
# Иногда одни и те же строчки 
# кода нужно повторить несколько раз без
# изменений кода (или с небольшими),
# тогда используют циклы
````

````python
a = 2

# Следующий код создаёт переменную i, 
# записывает в неё значение 0
# и проходит строчки со сложением и print(),
# каждый раз увеличивая i на 1. Последним значением i
# в этом цикле будет `4`
for i in range(5):
    a += 2
    print(a, i)
````

````python
# А так переменная i меняется от 2 
# до 6 (включительно)
for i in range(2, 7):
    print(i)
````

````python
# А так она меняется от 500 до 900 
# включительно с шагом 100
for i in range(500, 1000, 100):
    print(i)
````

````python
# Этот цикл вообще не запустится, 
# потому что правая граница 
# отрезка меньше левой
for i in range(50, 46):
    print(i)
````

````python
# А так запустится, если 
# сделать отрицательный шаг
for i in range(50, 46, -1):
    print(i)
````

````python
# Также в качестве пределов и шага можно
# использовать значения целочисленных перменных
start = 50
stop = 83
step = 7
for i in range(start, stop, step):
    print(i)
````

````python
# Ещё существует второй тип циклов,
# который исполняется до тех пор, пока
# условие не станет истинным

i = 2

while i < 5:
    print(i)
    i += 1
````

## 10_functions.py

````python
# Для многократного использования
# кода также используются функции

# Допустим, мы хотим посчитать 
# факториал двух чисел: 5! и 7!

# Так как алгоритм вычисления для 
# обоих чисел одинаков, то мы можем 
# записать этот алгоритм как функцию 
# от переменной n -- числа, от которого
# мы считаем факториал

# Здесь `fact` -- название функции
# `n` -- название переменной (может быть
# любым допустимым, как и название обычных
# переменных)
def fact(n):
    res = 1
    for i in range(1, n + 1):
        res *= i
    return res # значит что функция возвращает значение res
````

````python
g = fact(4)
print(g)
````

````python
print(fact(5))
print(fact(7))
````

## 12_oop.py

````python
# У некоторых типов помимо значения
# есть связанные с ними функции, 
# называемые методами. Иногда
# программисты используют их,
# чтобы сделать код проще.
# Подробнее обо всём этом вам расскажут 
# в весеннем семестре
````

````python
r = [0, 1, 2]

# Так можно добавить элемент
# в конец массива. Заметим, что
# написание немного отличается от вызова
# обычной функции
r.append(3)
print(r)
````

````python
arr = [9, 3, 2]
# Вот так можно отсортировать значения в массиве
arr.sort()
print(arr)
````

````python
# Теперь покажем, что тот же самый 
# результат можно получить, используя
# обычные функции

# Функция, производящая над
# массивом сортировку пузырьком
def my_sort(x):
    for i in range(len(x)):
        for j in range(len(x)):
            if x[j] >= x[i]:
                temp_variable = x[i]
                x[i] = x[j]
                x[j] = temp_variable
    return x
````

````python
# Теперь отсортируем массив `new_arr`
# (сравните количество символов в записи
# этого кода с сортировкой 
# массива arr выше)
new_arr = [5, 2, 6, 1]
sorted_arr = my_sort(new_arr)
print(sorted_arr)
````

````python
# Кортеж можно посторить и вручную,
# используя круглые скобки
tup = ('a', 'b', 'c', 0, 1.0, True)
print(tup[2], tup[4])
````

def double_factorial(n):
    remainder = n % 2
    res = 1
    for i in range(1, n + 1):
        if i % 2 == remainder:
            res *= i
    return res

print(double_factorial(a1) + double_factorial(a2) + double_factorial(a3))
